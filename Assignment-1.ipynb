```python
import numpy as np

X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)

print("\nEstimated model:")
print(f"Final Score = {beta[0]:.4f} + "
      f"({beta[1]:.4f} × Study Hours) + "
      f"({beta[2]:.4f} × Attendance) + "
      f"({beta[3]:.4f} × Assignment Score)")
```
